```python
# Two sample t-test
# To check whether the candidates who were hired and not hired have difference in the their projects count or not.
# (ie., to check whether the candidates are hired based on project count)

Hired = freshers[freshers["Hiring_Stage"] == "Offer"]["Projects_Count"]
Not_Hired = freshers[freshers["Hiring_Stage"] != "Offer"]["Projects_Count"]

print(f"Average project count of candidates who are hired : {Hired.mean()}")
print(f"Average project count of candidates who are not hired : {Not_Hired.mean()}")

from scipy.stats import ttest_ind

# H0 : There is no difference between the avrage projects count of candidates who are hired and not hired
# H1 : There is a significant difference between the average project count of candidates who are hired and not hired

t_statistic, p_value = ttest_ind(Hired, Not_Hired)
print(f"t_statistic : {t_statistic}")
print(f"p-value : {p_value}")

alpha = 0.05

if p_value < alpha:
    print("Reject H0, there is a differnce between the project count of candidates who are hired and not hired")
else:
    print("Failed to reject H0, There is no difference between the project count of candidates who are hired and not hired")

# so, companies didnt hire candidates only based on project count. Other factors may also influence the hiring outcome
```

```python
# Correlation
# To check whether the candidates with high CGPA gets high salary

CGPA_of_hired_candidates = freshers[freshers["Hiring_Stage"] == "Offer"]["CGPA"]
Salary = freshers[freshers["Hiring_Stage"] == "Offer"]["Offered_Salary_Inr"]

print(CGPA_of_hired_candidates)
print(Salary)

Salary = Salary.astype(float)
print(type(Salary))

from scipy import stats

#H0 : There is no relationship between CGPA and salary of the candidates
#H1 : There is a relationship between CGPA and salary of the candidates.

correlation, p_value = stats.pearsonr(CGPA_of_hired_candidates, Salary)
print(f"Correlation : {correlation}")
print(f"p-value : {p_value}")

#Correlation
if correlation > 0:
    print("There is a positive relationship")
elif correlation == 0:
    print("There is no relationship")
else:
    print("There is a negative relationship")

#P-value
alpha = 0.05

if p_value < alpha:
    print("Reject H0, There is a relationship between CGPA and salary of the candidates")
else:
    print("Failed to reject H0, There is no relationship between CGPA and salary of the candidates")
# There is a negatve relationship, so candidates with high salary gets low salary or the candidates with low salary gets high salary.
# But the relationship is very weak, so CGPA alone is not the reason for their high or low salary, Other factors may also influence the salary.
```

```python
# Probability 
# If I choose one candidate at random, what is the probability of choosing a candidate who has reference?

Reference = len(freshers[freshers["Referral_Applied"] == "Yes"])
print(Reference)
Total_candidates = len(freshers)
print(Total_candidates)

probability = Reference /Total_candidates
print(f"the probability of choosing a candidate who has reference is {round(probability * 100,2)}%")
```
